Logistic Regression

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
%matplotlib inline
```

```python
from sklearn.datasets import make_classification
```

```python
## create the dataset
X,y=make_classification(n_samples=1000,n_features=10,n_classes=2,random_state=46)
```

```python
pd.DataFrame(X)
```

```python
pd.DataFrame(y)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.30,random_state=46)
```

```python
## Model training
from sklearn.linear_model import LogisticRegression
logistic=LogisticRegression()
```

```python
logistic.fit(X_train,y_train)
```

```python
y_pred=logistic.predict(X_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
score=accuracy_score(y_test,y_pred)
cm=confusion_matrix(y_test,y_pred)
cr=classification_report(y_test,y_pred)
print(score)
print(cm)
print(cr)
```

**Hyperparameter Tuning** and **Cross Validation**

```python
model = LogisticRegression()
penalty=['l1','l2','elasticnet']
c_values=[100,10,1.0,0.1,0.01]
solver=['lbfgs', 'liblinear', 'newton-cg','sag', 'saga']
```

```python
params=dict(penalty=penalty,C=c_values,solver=solver)
```

```python
## Grid search cv
from sklearn.model_selection import GridSearchCV,StratifiedKFold
cv=StratifiedKFold()
grid = GridSearchCV(estimator=model,param_grid=params,scoring='accuracy',cv=cv,n_jobs=-1)
```

```python
grid
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred_grid=grid.predict(X_test)
```

```python
score=accuracy_score(y_test,y_pred_grid)
cm=confusion_matrix(y_test,y_pred_grid)
cr=classification_report(y_test,y_pred_grid)
print(score)
print(cm)
print(cr)
```

**Randomized SearchCV**

```python
from sklearn.model_selection import RandomizedSearchCV
model = LogisticRegression()
penalty=['l1','l2','elasticnet']
c_values=[100,10,1.0,0.1,0.01]
solver=['lbfgs', 'liblinear', 'newton-cg','sag', 'saga']
params=dict(penalty=penalty,C=c_values,solver=solver)
```

```python
randomCV=RandomizedSearchCV(estimator=model,param_distributions=params,cv=cv,n_jobs=-1,scoring='accuracy')
```

```python
randomCV.fit(X_train,y_train)
```

```python
randomCV.best_params_
```

```python
randomCV.best_score_
```

```python
y_pred_rcv=randomCV.predict(X_test)
score=accuracy_score(y_test,y_pred_rcv)
cm=confusion_matrix(y_test,y_pred_rcv)
cr=classification_report(y_test,y_pred_rcv)
print(score)
print(cm)
print(cr)
```

#### **Logistic Regression for Multiclass Classification Problem**

```python
X,y=make_classification(n_samples=1000,n_features=10,n_informative=4,n_classes=4,random_state=50)
```

```python
pd.DataFrame(X)
```

```python
pd.DataFrame(y)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.30,random_state=50)
```

```python
from sklearn.linear_model import LogisticRegression
logistic = LogisticRegression(multi_class='ovr')
logistic.fit(X_train,y_train)
y_pred=logistic.predict(X_test)
y_pred
```

```python
score=accuracy_score(y_test,y_pred)
cm=confusion_matrix(y_test,y_pred)
cr=classification_report(y_test,y_pred)
print(score)
print(cm)
print(cr)
```

```python
from sklearn.model_selection import RandomizedSearchCV
model = LogisticRegression()
penalty=['l1','l2','elasticnet']
c_values=[100,10,1.0,0.1,0.01]
solver=['lbfgs', 'liblinear', 'newton-cg','sag', 'saga']
params=dict(penalty=penalty,C=c_values,solver=solver)
randomCV=RandomizedSearchCV(estimator=model,param_distributions=params,cv=cv,n_jobs=-1,scoring='accuracy')
```

```python
randomCV.fit(X_train,y_train)
```

```python
randomCV.best_score_
```

```python
y_pred=randomCV.predict(X_test)
score=accuracy_score(y_test,y_pred)
cm=confusion_matrix(y_test,y_pred)
cr=classification_report(y_test,y_pred)
print(score)
print(cm)
print(cr)
```
