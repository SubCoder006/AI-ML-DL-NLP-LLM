Bernoulli Naive Bayes on "tips" dataset

```python
import pandas as pd
import seaborn as sns 
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import BernoulliNB
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

Load Dataset

```python
df = sns.load_dataset('tips')
df.head(5)
```

One-Hot encode 'sex' and 'smoker' columns

```python
df_encoded = df.copy()
```

```python
df_encoded['sex'] = df['sex'].map({'Male': 0, 'Female': 1}).astype(int)
df_encoded['smoker'] = df['smoker'].map({'No': 0, 'Yes': 1}).astype(int)
df_encoded.head(5)
```

target : predict 'time' column (Dinner or Lunch) based on other features
- encode target column 'time' to 0 and 1 using LabelEncoder

```python
df_encoded['time'] = df['time'].map({'Lunch':0,'Dinner':1}).astype(int)
```

```python
df_encoded.head(5)
```

##### BernoulliNB expects binary features, so binarize the numeric columns using their median as the threshold

```python
for col in ['total_bill','tip','size']:
    df_encoded[col] = (df_encoded[col]>df_encoded[col].median()).astype(int)
```

```python
df_encoded = pd.get_dummies(df_encoded, columns=['day'],drop_first=True)
df_encoded.head(5)
```

Split features (X) and target (y)

```python
X = df_encoded.drop('time', axis=1)
y = df_encoded['time']
```

```python
X.head(3)
```

```python
y.iloc[70:81]
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=50,stratify=y)
```

Train BernoulliNB model

```python
model = BernoulliNB()
model.fit(X_train,y_train)
```

Predict and evaluate

```python
y_pred = model.predict(X_test)

print(accuracy_score(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

Predict on test set and evaluate accuracy

```python
sample = X_test.iloc[10:21]
prediction = model.predict(sample)
print("Lunch : 0, Dinner : 1")
for i in prediction:
    print(i)
```
