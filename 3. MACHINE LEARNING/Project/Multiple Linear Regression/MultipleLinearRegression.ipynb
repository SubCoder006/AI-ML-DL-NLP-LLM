````python
# Lets prepare the dataframe
df=pd.DataFrame(dataset.data,columns=dataset.feature_names)
df
````

````python
df['Price']=dataset.target
df.head(10)
````

````python
X=df.iloc[:,:-1]   ## Independent 
y=df.iloc[:,-1]  ## Dependent
````

````python
X.head()
````

````python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.33,random_state=46)
````

````python
X_train.shape,X_test.shape
````

````python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
````

***MODEL TRANING***

````python
from sklearn.linear_model import LinearRegression
linreg = LinearRegression()
linreg.fit(X_train_scaled,y_train)
````

````python
linreg.coef_
````

````python
linreg.intercept_
````

````python
## Prediction for test data
y_pred=linreg.predict(X_test_scaled)
y_pred
````

````python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Basic metrics
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)
    
# Adjusted R2
n = len(y_test)
adj_r2 = 1 - (1 - r2) * (n - 1) / (n - X_train_scaled.shape[1] - 1)
    
# Print results
print(f"MAE  : {mae:.4f}")
print(f"MSE  : {mse:.4f}")
print(f"RMSE : {rmse:.4f}")
print(f"R2   : {r2:.4f}")
print(f"Adj R2: {adj_r2:.4f}")
    
````

#### **Pickling in Python**

##### 🔹 What is Pickling?

Pickling is the process of **serializing a Python object** into a byte stream so it can be:

* Saved to a file
* Transferred over a network
* Loaded later back into memory

The reverse process is called **Unpickling**.

---

##### 🔹 Core Idea

* **Pickle** → Convert object → Binary file
* **Unpickle** → Binary file → Original object

Handled using Python’s built-in `pickle` module.

---

##### 🔹 Why Use Pickling?

* Save trained Machine Learning models
* Store intermediate computation results
* Avoid recomputation
* Transfer objects between programs

---

##### 🔹 Basic Workflow

##### 1. Saving (Pickling)

```python
import pickle
data = {"model": "ridge", "accuracy": 0.98}
with open("data.pkl", "wb") as f:
    pickle.dump(data, f)
```

---

##### 2. Loading (Unpickling)

```python
import pickle
with open("data.pkl", "rb") as f:
    loaded_data = pickle.load(f)
print(loaded_data)
```

---

##### 🔹 Machine Learning Use Case

```python
#### Save model
with open("model.pkl", "wb") as f:
    pickle.dump(model, f)

#### Load model
with open("model.pkl", "rb") as f:
    model = pickle.load(f)

#### Predict
y_pred = model.predict(X_test)
```

---

##### 🔹 Key Points

* Supports most Python objects (lists, dictionaries, models, etc.)
* Common file extension: `.pkl`
* Stored in binary format (not human-readable)
* Faster than JSON for complex objects

---

##### 🔹 ⚠️ Security Warning

Never load pickle files from **untrusted sources**.
Pickle can execute arbitrary code during unpickling.

---

##### 🔹 Alternative: Joblib

Better suited for large NumPy arrays and ML models.

```python
import joblib

joblib.dump(model, "model.pkl")
model = joblib.load("model.pkl")
```
