```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_samples, silhouette_score
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')
%matplotlib inline
```

```python
X,y = load_iris(return_X_y=True)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X_scaled
```

```python
y
```

```python
plt.scatter(X[:,0],X[:,1],X[:,2],X[:,3])
```

```python
kmeans = KMeans()
```

```python
silhouette_coefficients=[]
for k in range(2,7):
    kmeans = KMeans(n_clusters=k,init='k-means++',random_state=42,n_init=10)
    kmeans.fit(X_scaled)
    score = silhouette_score(X_scaled,kmeans.labels_)
    silhouette_coefficients.append(score)
```

```python
silhouette_coefficients
```

```python
## plotting silhouette score
plt.plot(range(2,7),silhouette_coefficients)
plt.xticks(range(2,7))
plt.xlabel("Number of Cluters")
plt.ylabel("Silhoutte Coeffecient")
plt.show()
```

```python
# best at k = 2
kmeans = KMeans(n_clusters=2,n_init=10,init='k-means++',random_state=42)
kmeans.fit(X_scaled)
labels = kmeans.labels_
```

```python
overall_score = silhouette_score(X_scaled,labels)
sample_scores = silhouette_samples(X_scaled,labels)
```

```python
overall_score
```

```python
sample_scores
```

```python
plt.figure(figsize=(15,10))
y_lower = 10
for i in range(2):
    cluster_vals = sample_scores[labels == i]
    cluster_vals.sort()
    size = cluster_vals.shape[0]
    y_upper = y_lower + size

    plt.fill_betweenx(np.arange(y_lower, y_upper), 0, cluster_vals, alpha=0.7, label=f'Cluster {i}')
    y_lower = y_upper + 10   # gap between clusters

plt.axvline(x=overall_score, color='red', linestyle='--', label='Avg score')
plt.xlabel("Silhouette coefficient s(i)")
plt.ylabel("Points (grouped by cluster)")
plt.title(f"Silhouette Plot (k={2}, avg={overall_score:.3f})")
plt.legend()
plt.tight_layout()
plt.show()
```
